```python
#GCD 2 types
from random import randint

def GCD(a, b):
    while b:
        a,b = b,a%b
    return a
def GCD_binary(a,b):
    g =1
    while a%2 ==0 & b%2 ==0:
        a//=2
        b//=2
        g*=2
    while a!= 0:
        while a%2 ==0: a//=2
        while b%2 ==0: b//=2
        if a>= b: a = (a-b)//2
        else: b = (b-a)//2
    return g*b
def extended_euclid(a,b):
    s, old_s,t,old_t,r, old_r = 0,1,1,0,b,a
    while r != 0:
        q = round(old_r/r)
        old_r, r = r, old_r - q*r
        old_s, s = s, old_s - q*s
        old_t, t = t, old_t - q*t
    d,x,y = old_r, t,s
    return d,x,y
def inv_mod(a,m):
    d,x,y = extended_euclid(a,m)
    if d!= 1:
        return None
    else:
        return x%m
def CRT(a,b):
    #a = [divisors], b = [remainders]
    M = 1
    for i in b:
        M*=i
    x=0 
    Mi = []
    for i in b:
        Mi.append(M//i)
    yi = []
    for i in Mi:
        yi.append(inv_mod(i,M//i))
    for i in range(len(a)):
        x+= pow(a[i]*Mi[i]*yi[i],1,M)
    return x%M
```
